```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/StudentsPerformance (1).csv")

df.head()
```

```python
A = np.array([
    df.loc[0, 'math score'],
    df.loc[0, 'reading score'],
    df.loc[0, 'writing score']
])

print(A)
```

```python
# Norm-1

norm1 = np.linalg.norm(A, 1)

print("Norm-1 =", norm1)
```

```python
# Norm-2

norm2 = np.linalg.norm(A, 2)

print("Norm-2 =", norm2)
```

```python
B = np.array([
    df.loc[1, 'math score'],
    df.loc[1, 'reading score'],
    df.loc[1, 'writing score']
])
```

```python
# DOT Product

dot_product = np.dot(A, B)
print(dot_product)
```

```python
angle = np.arccos(dot_product / (np.linalg.norm(A) * np.linalg.norm(B)))
print(angle)
```

```python
# Find cross product A×B

cross_product = np.cross(A, B)
print(cross_product)
```

```python
projection = (np.dot(A, B) / np.dot(B, B)) * B
print(projection)
```

```python
matrix = df[['math score', 'reading score', 'writing score']].head(5)
M = matrix.values
```

```python
# ADDITION

addition = M + M
print(addition)
```

```python
# MULTIPICATION

multiplication = np.dot(M, M.T)
print(multiplication)
```

```python
# Transpose
transpose = M.T
print(transpose)
```

```python
# inverse
square_matrix = M[:3, :3]
inverse = np.linalg.inv(square_matrix)
print(inverse)
```

```python
# Determinant
determinant = np.linalg.det(square_matrix)
print(determinant)
```

PART D : Eigenvalues & Decomposition

```python
# Covariance Matrix

data = df[['math score', 'reading score', 'writing score']]
cov_matrix = np.cov(data.T)
print(cov_matrix)
```

```python
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)
print(eigenvalues)
print(eigenvectors)
```

```python
# LU decomposition

from scipy.linalg import lu

matrix = df[['math score', 'reading score', 'writing score']].head(3).values

P, L, U = lu(matrix)
print(P)
print(L)
print(U)
```

```python
# Singular Value Decomposition (SVD)

U, S, VT = np.linalg.svd(data)
print(U)
print(S)
print(VT)
```

```python
# PCA

from sklearn.decomposition import PCA

X = df[['math score', 'reading score', 'writing score']]
pca = PCA(n_components=2)
principal_components = pca.fit_transform(X)
print(principal_components[:5])
plt.scatter(principal_components[:,0],principal_components[:,1])
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.show()
```

```python
df['average'] = (df['math score'] + df['reading score'] + df['writing score']) / 3
```

```python
df['category'] = np.where(df['average'] >= 70,'Above Average','Below Average')
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
y = le.fit_transform(df['category'])
```

```python
# LDA

from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

X = df[['math score', 'reading score', 'writing score']]
lda = LinearDiscriminantAnalysis(n_components=1)
X_lda = lda.fit_transform(X, y)
print(X_lda[:5])

plt.scatter(X_lda,np.zeros_like(X_lda))
plt.show()
```
